# ESP32 Thermal-Risk TinyML — Teaching Notebook

This is a **teaching prototype**, not a certified fire detector. It uses DHT22 temperature and humidity only. The output is `NORMAL`, `ELEVATED_THERMAL_RISK`, or `HIGH_THERMAL_RISK`. The flame sensor remains a separate direct confirmation signal.

We use real room-baseline data plus synthetic, smooth heat-risk patterns so students can see the full TinyML workflow before MQ-2 data is available. Do not present synthetic accuracy as real fire-detection accuracy.

## 1. Setup

Run locally in Jupyter/VS Code, or upload this notebook and CSV files to Google Colab. Required packages: `numpy pandas matplotlib scikit-learn tensorflow`. TensorFlow is only needed for the deployable tiny dense model.

Collect one or more ESP32 exports into `data/`, for example `data/room_a.csv`. Each must contain: `sequence,temp_c,humidity_pct`.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

SAMPLE_SECONDS = 2
WINDOW_SAMPLES = 10              # 20 seconds of DHT history
FORECAST_HORIZON_SECONDS = 60    # teaching target: risk in the next minute
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

LABELS = ['NORMAL', 'ELEVATED_THERMAL_RISK', 'HIGH_THERMAL_RISK']
FEATURE_COLUMNS = ['temp_c', 'humidity_pct']
ARTIFACT_DIR = Path('../artifacts')
ARTIFACT_DIR.mkdir(exist_ok=True)
```

## 2. Load real baseline data

Use several locations and normal conditions: room, kitchen with no cooking, near a window, fan on/off. These are real `NORMAL` examples. Keep future controlled heat-source sessions in separate CSV files so they can become an honest final test set.

```python
data_files = sorted(Path('../data').glob('*.csv'))
if not data_files:
    raise FileNotFoundError('Add new ESP32 DHT exports to ../data first.')

sessions = []
for file_path in data_files:
    frame = pd.read_csv(file_path)
    missing = set(['sequence', *FEATURE_COLUMNS]) - set(frame.columns)
    if missing:
        print(f'Skipping {file_path.name}: missing {missing}')
        continue
    frame = frame[['sequence', *FEATURE_COLUMNS]].dropna().copy()
    frame['session'] = file_path.stem
    sessions.append(frame)

if not sessions:
    raise ValueError('No compatible DHT-only CSV files found.')

baseline = pd.concat(sessions, ignore_index=True)
display(baseline.head())
print('Rows:', len(baseline), '| sessions:', baseline.session.nunique())
baseline.groupby('session')[FEATURE_COLUMNS].agg(['min', 'max', 'mean'])
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 5), sharex=True)
for session_name, frame in baseline.groupby('session'):
    axes[0].plot(frame['sequence'], frame['temp_c'], label=session_name)
    axes[1].plot(frame['sequence'], frame['humidity_pct'], label=session_name)
axes[0].set_ylabel('Temperature C')
axes[1].set_ylabel('Humidity %')
axes[1].set_xlabel('ESP32 sequence')
axes[0].legend(loc='best')
plt.show()
```

## 3. Make synthetic teaching labels

A synthetic window starts from a real 20-second baseline segment. We add a smooth thermal curve rather than random extreme values. `ELEVATED` means a gentle upward trend; `HIGH` means a stronger sustained trend. These labels stand for **simulated risk in the next 60 seconds**, not a claim that a real flame will appear.

```python
def consecutive_windows(frame):
    values = frame[FEATURE_COLUMNS].to_numpy(dtype=np.float32)
    if len(values) < WINDOW_SAMPLES:
        return np.empty((0, WINDOW_SAMPLES, len(FEATURE_COLUMNS)), dtype=np.float32)
    return np.stack([values[i:i + WINDOW_SAMPLES] for i in range(len(values) - WINDOW_SAMPLES + 1)])

def add_synthetic_risk(window, label):
    result = window.copy()
    progress = np.linspace(0.0, 1.0, WINDOW_SAMPLES, dtype=np.float32)
    if label == 1:       # early trend, expected elevated risk
        temp_rise = np.random.uniform(0.25, 0.70)
        humidity_shift = np.random.uniform(-1.5, 0.2)
    elif label == 2:     # stronger trend, expected high risk
        temp_rise = np.random.uniform(0.80, 2.00)
        humidity_shift = np.random.uniform(-4.0, -0.5)
    else:
        return result
    curve = progress ** 1.7
    result[:, 0] += temp_rise * curve
    result[:, 1] += humidity_shift * curve
    result += np.random.normal(0, [0.03, 0.08], result.shape).astype(np.float32)
    return result

window_sets, session_groups = [], []
for session_name, session_frame in baseline.groupby('session'):
    windows = consecutive_windows(session_frame)
    if len(windows):
        window_sets.append(windows)
        session_groups.extend([session_name] * len(windows))
base_windows = np.concatenate(window_sets) if window_sets else np.empty((0, WINDOW_SAMPLES, 2))
if len(base_windows) == 0:
    raise ValueError('Need at least 10 DHT rows in one session.')

X_windows = np.concatenate([
    base_windows,
    np.stack([add_synthetic_risk(w, 1) for w in base_windows]),
    np.stack([add_synthetic_risk(w, 2) for w in base_windows]),
])
y = np.concatenate([
    np.zeros(len(base_windows), dtype=np.int8),
    np.ones(len(base_windows), dtype=np.int8),
    np.full(len(base_windows), 2, dtype=np.int8),
])
X = X_windows.reshape(len(X_windows), -1)
groups = np.tile(np.asarray(session_groups), 3)
print('Feature shape:', X.shape, '| class counts:', np.bincount(y))
```

## 4. Train an explainable baseline

Train/test splits are by complete recording session. Synthetic variants derived from a session stay on the same side of the split. Collect at least two baseline sessions before training.

```python
if len(set(groups)) < 2:
    raise ValueError('Collect at least two separate baseline sessions before training.')
splitter = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=RANDOM_SEED)
train_index, test_index = next(splitter.split(X, y, groups))
X_train, X_test = X[train_index], X[test_index]
y_train, y_test = y[train_index], y[test_index]
scaler = StandardScaler().fit(X_train)
baseline_model = LogisticRegression(max_iter=1000, multi_class='multinomial', random_state=RANDOM_SEED)
baseline_model.fit(scaler.transform(X_train), y_train)
predicted = baseline_model.predict(scaler.transform(X_test))
print(classification_report(y_test, predicted, target_names=LABELS))
ConfusionMatrixDisplay.from_predictions(y_test, predicted, display_labels=LABELS, xticks_rotation=20)
plt.show()
```

## 5. Tiny dense model and INT8 export

Run this after the baseline works. The representative data is from training only. The generated `.tflite` and normalization JSON are inputs for the later ESP32 inference sketch.

```python
import tensorflow as tf

tiny_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train.shape[1],)),
    tf.keras.layers.Dense(12, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax'),
])
tiny_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history = tiny_model.fit(
    scaler.transform(X_train), y_train,
    validation_split=0.2, epochs=40, batch_size=32, verbose=0
)
loss, accuracy = tiny_model.evaluate(scaler.transform(X_test), y_test, verbose=0)
print(f'Tiny model test accuracy: {accuracy:.3f} (synthetic teaching metric)')

converter = tf.lite.TFLiteConverter.from_keras_model(tiny_model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
def representative_data():
    for row in scaler.transform(X_train[:200]).astype(np.float32):
        yield [row.reshape(1, -1)]
converter.representative_dataset = representative_data
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

model_bytes = converter.convert()
model_path = ARTIFACT_DIR / 'thermal_risk_model_int8.tflite'
model_path.write_bytes(model_bytes)
(ARTIFACT_DIR / 'thermal_risk_normalization.json').write_text(json.dumps({
    'feature_order': [f'{field}_{i}' for i in range(WINDOW_SAMPLES) for field in FEATURE_COLUMNS],
    'mean': scaler.mean_.tolist(), 'scale': scaler.scale_.tolist(),
    'labels': LABELS, 'sample_seconds': SAMPLE_SECONDS,
    'forecast_horizon_seconds': FORECAST_HORIZON_SECONDS,
}, indent=2))
print('Saved', model_path, '| bytes:', len(model_bytes))
```

## Before calling it a fire model

Repeat collection with controlled heat/candle sessions, use real event logs for labels, and test on unseen sessions. Then add MQ-2 to the input. Flame detection should remain a direct alert override, not a prediction target.
